## 2. Creating a Sample Retail `DataFrame`
A Pandas `DataFrame` is a two-dimensional, size-mutable, tabular data structure with labeled axes (rows and columns). It is the standard structure for exploratory data analysis (EDA).

Below, we create a synthetic sample retail dataset representing 10 store outlets across key retail corridors in Bengaluru (e.g., Koramangala, Indiranagar, Whitefield, HSR Layout, Jayanagar).

## 5. Statistical Profiling using `describe()`
The `.describe()` method generates descriptive statistics summarizing the central tendency, dispersion, and shape of a dataset's distribution:
- `count`: Non-missing values.
- `mean` & `std`: Average value and degree of variation.
- `min`, `25%` (Q1), `50%` (median / Q2), `75%` (Q3), `max`: Five-number summary for detecting skewness and potential outliers.

We also inspect categorical distribution using `describe(include=['object'])`.

```python
# Descriptive statistics for numerical variables
print("--- Summary Statistics for Numerical Metrics (describe()) ---")
display(df_retail.describe().round(2))

# Summary statistics for categorical variables
print("--- Summary Statistics for Categorical Fields (describe(include=['object'])) ---")
display(df_retail.describe(include=['object']))
```

## 6. Aggregations and Totals using `sum()`
The `.sum()` method computes the sum of numerical values.

**Applications for GeoPulse Analytics:**
- Calculating high-level retail KPIs: Total Footfall, Total Transactions, and Gross Revenue.
- Calculating localized sums via grouped aggregation (`groupby().sum()`) to compare footfall and revenue across retail micro-markets (e.g., Koramangala vs. Indiranagar).

```python
# 1. High-level metric sums
total_footfall = df_retail['footfall_count'].sum()
total_transactions = df_retail['transactions'].sum()
total_revenue = df_retail['daily_sales_inr'].sum()

print("=" * 55)
print("High-Level Retail Aggregate Metrics (sum())")
print("=" * 55)
print(f"Total Retail Footfall    : {total_footfall:,} visitors")
print(f"Total Transactions       : {total_transactions:,} transactions")
print(f"Gross Daily Sales Revenue: INR {total_revenue:,.2f}")
print("=" * 55)

# 2. Multi-column numerical sums
print("\n--- Column-wise Numerical Sums ---")
numeric_sums = df_retail[['footfall_count', 'transactions', 'daily_sales_inr']].sum()
display(numeric_sums)

# 3. Location-wise grouping and summation
print("\n--- Location-wise Footfall and Revenue Totals (groupby + sum) ---")
location_aggregates = (
    df_retail.groupby('location')[['footfall_count', 'transactions', 'daily_sales_inr']]
    .sum()
    .reset_index()
    .rename(columns={
        'footfall_count': 'total_footfall',
        'transactions': 'total_transactions',
        'daily_sales_inr': 'total_sales_inr'
    })
)

# Format revenue for presentation
location_aggregates['formatted_sales'] = location_aggregates['total_sales_inr'].apply(lambda x: f"INR {x:,.2f}")
display(location_aggregates)
```
